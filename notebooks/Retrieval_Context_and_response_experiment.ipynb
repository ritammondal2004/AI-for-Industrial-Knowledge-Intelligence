# ------ Continuous RAG Chat Bot with **MMR**----------

````python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import ChatPromptTemplate

# retrival model
retriev_model = ChatGoogleGenerativeAI(
    model="gemini-2.5-flash",
    temperature=0.2,
    google_api_key=os.environ["GOOGLE_API_KEY"]
)

# ------fixes the phrasing sensitivity problem-----
EXPANSION_PROMPT = ChatPromptTemplate.from_template("""
You are a query expansion assistant for an industrial knowledge base.
Given a user question, generate 3 alternative phrasings that mean the same thing
but use different vocabulary. Industrial/technical documents often use formal
terminology, so include both casual and technical phrasings.

Return ONLY a JSON list of 3 strings, nothing else.
Example: ["What is cavitation?", "Define cavitation in pumps", "Cavitation phenomenon explanation"]

Question: {question}
""")

````

````python
import re
import json
from langchain_core.output_parsers import StrOutputParser

expansion_chain = EXPANSION_PROMPT | retriev_model | StrOutputParser()

def expand_query(query):
    """Generate alternative phrasings to improve retrieval recall."""
    try:
        raw = expansion_chain.invoke({"question": query})
        raw = raw.strip()
        if raw.startswith("```"):
            raw = raw.split("```")[1]
            if raw.startswith("json"):
                raw = raw[4:]
        alternatives = json.loads(raw.strip())
        all_queries = [query] + alternatives[:3]
        return all_queries
    except Exception as e:
        print(f"Query expansion failed: {e} — using original query only")
        return [query]
````

````python
# ── Metadata-based folder routing -------

FOLDER_KEYWORDS = {
    "Regulations": [
        "osha", "regulation", "standard", "compliance", "legal",
        "oisd", "factory act", "peso", "statutory", "dgms", "mandatory","CPL"
    ],
    "Incident_Reports": [
        "incident", "accident", "failure", "fire", "explosion",
        "injury", "fatality", "near miss", "root cause", "investigation"
    ],
    "Manuals": [
        "install", "operation", "specification", "dimension", "part number",
        "assembly", "disassembly", "torque", "oem", "technical data"
    ],
    "Procedures": [
        "procedure", "sop", "step", "permit", "lockout", "tagout",
        "startup", "shutdown", "checklist", "how to"
    ]
}

def detect_relevant_folders(query):
    """
    Returns list of folders to boost/filter based on query keywords.
    Returns None if no strong signal — means search everything.
    """
    query_lower = query.lower()
    folder_scores = {}

    for folder, keywords in FOLDER_KEYWORDS.items():
        score = sum(1 for kw in keywords if kw in query_lower)
        if score > 0:
            folder_scores[folder] = score

    if not folder_scores:
        return None  # no signal — search all folders

    # Return folders that scored at least half the max score
    max_score = max(folder_scores.values())
    relevant = [f for f, s in folder_scores.items() if s >= max_score * 0.5]
    return relevant if relevant else None
````

#### ── Smart retriever with MMR + expansion + metadata routing ─────

````python


def smart_retrieve(query, k_final=5):
    """
    Full retrieval pipeline:
    1. Expand query into multiple phrasings
    2. Detect relevant folders from query keywords
    3. Retrieve with MMR (reduces duplicate chunks)
    4. Merge + deduplicate across all query variants
    5. Return top k_final unique chunks
    """
    expanded_queries = expand_query(query)
    relevant_folders = detect_relevant_folders(query)

    if relevant_folders:
        print(f"  → Folder routing: {relevant_folders}")
    else:
        print(f"  → No folder signal — searching all folders")

    seen_ids = set()
    all_docs = []

    for q in expanded_queries:
        try:
            # MMR retrieval: fetch more candidates, then diversify
            if relevant_folders:
                # Metadata filter — only search relevant folders
                if len(relevant_folders) == 1:
                    filter_dict = {"folder": relevant_folders[0]}
                else:
                    filter_dict = {"folder": {"$in": relevant_folders}}

                docs = chroma_db.max_marginal_relevance_search(
                    q,
                    k=k_final,
                    fetch_k=k_final * 3,  # fetch 3x, MMR picks best diverse subset
                    filter=filter_dict
                )
            else:
                docs = chroma_db.max_marginal_relevance_search(
                    q,
                    k=k_final,
                    fetch_k=k_final * 3
                )

            for doc in docs:
                # Deduplicate by content hash
                doc_id = hash(doc.page_content[:100])
                if doc_id not in seen_ids:
                    seen_ids.add(doc_id)
                    all_docs.append(doc)

        except Exception as e:
            print(f"  Retrieval failed for query variant '{q[:50]}': {e}")
            continue

    # Return top k_final after de-duplication across all query variants
    return all_docs[:k_final]
````

### ── Citation formatting — structured source list at end ─────────

````python


def format_docs_with_citations(docs):
    return "\n\n".join(doc.page_content for doc in docs)



#########
def format_sources_block(docs):
    """
    Builds a clean de-duplicated source list for display below the answer.
    Groups by filename, lists all pages referenced.
    """
    sources = {}
    for doc in docs:
        fname  = doc.metadata.get('source', 'unknown')
        page   = doc.metadata.get('page', '?')
        folder = doc.metadata.get('folder', 'unknown')

        if fname not in sources:
            sources[fname] = {"pages": set(), "folder": folder}

        # Only add if not already present — deduplicates same page from multiple variants
        sources[fname]["pages"].add(str(page))

    lines = ["\n📚 **Sources Referenced:**"]
    for fname, info in sorted(sources.items()):   # sorted for consistent ordering
        pages = ", ".join(
            sorted(info["pages"],
                   key=lambda x: int(x) if x.isdigit() else 0)
        )
        lines.append(f"  • `{fname}` — Page(s): {pages}  |  {info['folder']}")

    return "\n".join(lines)

#########
def format_history(history):
    if not history:
        return "No previous conversation."
    return "\n\n".join(
        f"Human: {q}\nAssistant: {a}"
        for q, a in history
    )
````

### ── Updated ask()

````python
chat_model = ChatGoogleGenerativeAI(
    model="gemini-2.5-flash",
    temperature=0.5,
    google_api_key=os.environ["GOOGLE_API_KEY"]
)
````

````python

conversational_prompt = ChatPromptTemplate.from_template(""".
You are an Industrial Knowledge Assistant specialized in heavy industry operations,
equipment maintenance, safety procedures, incident reports and regulatory compliance.

Answer using ONLY the retrieved context below.
Cite source and page for every fact: [Source: filename, Page X]
If context is insufficient, 1st try to find the semantic meaning in that and answer
if extremely missing say so clearly, explain your problem to the user. Do not hallucinate.

--- Retrieved Context ---
{context}

--- Recent Conversation History ---
{chat_history}

--- Current Question ---
{question}

Answer:
""")


def ask(query, chat_history):

    # Smart retrieval: expansion + MMR + folder routing
    docs = smart_retrieve(query, k_final=5)

    if not docs:
        return "I couldn't find relevant information in the knowledge base for that query.", chat_history

    context = format_docs_with_citations(docs)
    sources_block = format_sources_block(docs)
    recent_history = chat_history[-3:]
    history_text = format_history(recent_history)

    RAGchain = conversational_prompt | chat_model | StrOutputParser()
    answer = RAGchain.invoke({
        "context": context,
        "question": query,
        "chat_history": history_text
    })

    # Append source block below the answer
    full_response = answer + "\n\n" + sources_block

    return full_response, chat_history + [(query, answer)]
````

````python
# ── smart_retrieve — called directly from ask() ───────────────────

def smart_retrieve(query, k_final=5, verbose=False):
    """
    Full retrieval pipeline:
    1. Expand query
    2. Detect relevant folders
    3. MMR retrieval per variant
    4. Deduplicate across variants
    5. Return top k_final unique docs
    """
    expanded_queries   = expand_query(query)
    relevant_folders   = detect_relevant_folders(query)

    if verbose: 
        print(f"  Expanded to {len(expanded_queries)} variants")
        if relevant_folders:
            print(f"  Folder routing → {relevant_folders}")
        else:
            print(f"  No folder signal — searching all folders")

    seen_ids  = set()
    all_docs  = []

    for q in expanded_queries:
        try:
            if relevant_folders:
                filter_dict = (
                    {"folder": relevant_folders[0]}
                    if len(relevant_folders) == 1
                    else {"folder": {"$in": relevant_folders}}
                )
                docs = chroma_db.max_marginal_relevance_search(
                    q, k=k_final, fetch_k=k_final * 3,
                    filter=filter_dict
                )
            else:
                docs = chroma_db.max_marginal_relevance_search(
                    q, k=k_final, fetch_k=k_final * 3
                )

            for doc in docs:
                doc_id = hash(doc.page_content[:100])
                if doc_id not in seen_ids:
                    seen_ids.add(doc_id)
                    all_docs.append(doc)

        except Exception as e:
            if verbose:
                print(f"  Retrieval failed for variant '{q[:40]}': {e}")
            continue

    return all_docs[:k_final]
````

####  Graph Retrieval Function

````python
# ── Graph context retrieval ─────────────
def graph_context(query, G, top_k=5, max_neighbors=4):
    """
    1. Extract meaningful keywords from query
    2. Find matching nodes (case-insensitive partial match)
    3. Expand to immediate neighbors
    4. Return concise textual summary

    Args:
        query      : user question string
        G          : NetworkX DiGraph
        top_k      : max number of seed nodes to match
        max_neighbors : max neighbors to expand per seed node

    Returns:
        str : formatted graph context summary, or "" if nothing found
    """
    if G.number_of_nodes() == 0:
        return ""

    # ── Step 1: extract keywords (skip stopwords) ──────────
    STOPWORDS = {
        "what", "how", "why", "when", "where", "which", "who",
        "is", "are", "was", "were", "the", "a", "an", "in",
        "of", "to", "for", "on", "and", "or", "tell", "me",
        "about", "explain", "describe", "does", "do", "can",
        "should", "would", "give", "list", "show"
    }
    keywords = [
        w.strip("?.,:;")
        for w in query.lower().split()
        if len(w) > 3 and w.lower() not in STOPWORDS
    ]

    if not keywords:
        return ""

    # ── find matching seed nodes
    matched_nodes = []
    for node in G.nodes():
        node_lower = node.lower()
        if any(kw in node_lower for kw in keywords):
            matched_nodes.append(node)

    matched_nodes = matched_nodes[:top_k]

    if not matched_nodes:
        return ""

    # ──expand to neighbors ─
    context_lines = []

    for node in matched_nodes:
        attrs  = G.nodes[node]
        node_type  = attrs.get("type", "Unknown")
        description = attrs.get("description", "")
        sources  = attrs.get("sources", [])
        source_str  = sources[0].get("source_pdf", "?") if sources else "?"

        # Node summary line
        context_lines.append(
            f"Entity: {node} | Type: {node_type} | Source: {source_str}"
        )
        if description:
            context_lines.append(f"  Description: {description[:120]}")

        # Outgoing relationships
        successors = list(G.successors(node))[:max_neighbors]
        for tgt in successors:
            rels = G[node][tgt].get("relations", ["related_to"])
            context_lines.append(f"  → [{rels[0]}] → {tgt}")

        # Incoming relationships
        predecessors = list(G.predecessors(node))[:max_neighbors]
        for src in predecessors:
            rels = G[src][node].get("relations", ["related_to"])
            context_lines.append(f"  ← {src} [{rels[0]}] ←")

        context_lines.append("")  # blank line between nodes

    return "\n".join(context_lines).strip()
````

#### Gemini Call With API Key Rotation
 We are using multiple gemini API keys. We did this so if our free tire API limit hit , it will automatically switch to available API key

````python
# ── Gemini generation with automatic key rotation + retry
# Handles 429 / quota exceeded gracefully.

from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.output_parsers import StrOutputParser

def get_langchain_model(temp=0.2):
    """Returns a fresh LangChain ChatGoogleGenerativeAI using current key."""
    return ChatGoogleGenerativeAI(
        model="gemini-2.5-flash",
        temperature=temp,
        google_api_key=API_KEYS[current_key]
    )

def invoke_with_rotation(chain_input, max_retries=None):
    """
    Invokes the RAG chain with automatic API key rotation on quota errors.

    Rotation logic:
    - On 429 / quota exceeded → switch key → retry immediately
    - If all keys exhausted    → wait 40 seconds → retry from key 0
    - Repeats until success or max_retries reached
    """
    global current_key

    if max_retries is None:
        max_retries = len(API_KEYS) *2  # two full rotations before giving up

    attempts = 0
    keys_exhausted_count = 0

    while attempts < max_retries:
        try:
            # Rebuild model with current key each attempt
            chat_model = get_langchain_model(0.6)
            chain = conversational_prompt_enhanced | chat_model | StrOutputParser()
            return chain.invoke(chain_input)

        except Exception as e:
            err_str = str(e).lower()

            if "429" in err_str or "quota" in err_str or "exhausted" in err_str or "503" in err_str:
                next_key_idx = (current_key + 1) % len(API_KEYS)

                switch_api_key()
                attempts += 1
                continue

            else:
                # Non-quota error — don't retry, raise immediately
                raise e

    raise RuntimeError(
        f"All {max_retries} retry attempts failed across {len(API_KEYS)} API keys."
    )

print("Gemini rotation-aware invoker ready.")
````

####  Updated ask() — Hybrid GraphRAG

````python
# ── Hybrid GraphRAG ask() — uses smart_retrieve directly ─────

def ask(query, chat_history, verbose=False):
    """
    Hybrid GraphRAG single turn.
    Returns (answer_with_sources, updated_history)
    """

    # 1. Vector retrieval via smart_retrieve ─────
    all_docs = smart_retrieve(query, k_final=5, verbose=verbose)

    # 2. Knowledge Graph retrieval
    kg_ctx = graph_context(query, G, top_k=5, max_neighbors=4)
    if verbose:
        if kg_ctx:
            print(f"  Graph context: {len(kg_ctx.splitlines())} lines")
        else:
            print(f"  Graph context: no matching entities found")

    # ─Format vector context + sources ────
    if all_docs:
        vector_ctx    = format_docs_with_citations(all_docs)
        sources_block = format_sources_block(all_docs)
    else:
        vector_ctx    = "No relevant document chunks retrieved."
        sources_block = ""

    # ── Conversation memory
    history_text = format_history(chat_history[-3:])

    # ──Gemini generation with key rotation
    answer = invoke_with_rotation({
        "context"      : vector_ctx,
        "graph_context": kg_ctx if kg_ctx else "No relevant graph entities found.",
        "chat_history" : history_text,
        "question"     : query,
    })

    # ── 6. Append sources below answer ───────────────────────────
    full_response = answer
    if sources_block:
        full_response += "\n\n" + sources_block

    return full_response, chat_history + [(query, answer)]
````
